```python
import numpy as np
def m_to_km(elevation):
    """
    Converts elevation measurements from meters to kilometers.

    This function handles missing topographical data by checking for 
    null values (NaN) before performing the division. This prevents 
    calculation errors in environmental modeling datasets.

    Args:
        elevation (float or int): The altitude/elevation value in meters.
            Can also be a np.nan or None.

    Returns:
        float: The elevation converted to kilometers (elevation / 1000).
        np.nan: If the input elevation was missing or null.

    Example:
        >>> m_to_km(1500)
        1.5
        >>> m_to_km(None)
        nan
    """
    if pd.isna(elevation): #If it is null
        return np.nan
    return elevation / 1000
```

```python
m_to_km(28)
```

```python
#Convert everything in the Elev column from meters to km
#Using .apply() 
elev_km = sel_df["Elev"].apply(m_to_km)
elev_km
```

```python
#Replace the old values of Elev with the new ones in the "sel_df" data frame
sel_df["Elev"] = sel_df["Elev"].apply(m_to_km)
```

```python
sel_df
```

```python
#Create a new column by the sum of Lat and Long
def lat_sum_long(lat, long):
    """
    Sums up the value of the latitude and longitude in each row.

    This function deals with missing lat and long values
    by checking for Null values (NaN) before carrying out the sum operation
    and ensures that calculation errors are prevented

    Args:
        Latitude and longitude values (float or integer)
            Can also be a np.nan or None.
    
    Returns:
        float: sum of lat and long values
        np.nan: If the lat or long value was missing or null.
    
    Example:
        >>> lat_sum_long(2, 5)
        7
        >>> lat_sum_long(None, None)
        nan
    """
    if pd.isna(lat) or pd.isna(long):
        return np.nan
    return lat + long
```

```python
sel_df["sum_coords"] = sel_df.apply(lambda row: lat_sum_long(row.TH_LAT, row.TH_LONG), axis = 1)
```
